```python
subscriptions = pd.read_csv("ravenstack_subscriptions.csv")
feature_usage = pd.read_csv("ravenstack_feature_usage.csv")
support_tickets = pd.read_csv("ravenstack_support_tickets.csv")
churn_events = pd.read_csv("ravenstack_churn_events.csv")
```

```python
tables = {
    "Subscriptions": subscriptions,
    "Feature Usage": feature_usage,
    "Support Tickets": support_tickets,
    "Churn Events": churn_events
}

for name, df in tables.items():
    print(f"\n--- {name} ---")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
```

```python
tables = {
    "Accounts": accounts,
    "Subscriptions": subscriptions,
    "Feature Usage": feature_usage,
    "Support Tickets": support_tickets,
    "Churn Events": churn_events
}

for name, df in tables.items():
    print(f"\n{'=' * 50}")
    print(f"{name}")
    print(f"{'=' * 50}")

    print("Shape:", df.shape)
    print("\nMissing values:")
    print(df.isnull().sum())

    print("\nDuplicate rows:", df.duplicated().sum())

    print("\nUnique values:")
    print(df.nunique())
```

```python
accounts["signup_date"] = pd.to_datetime(accounts["signup_date"])
```

```python
accounts["signup_date"].dtype
```

```python
subscriptions["start_date"] = pd.to_datetime(subscriptions["start_date"])
subscriptions["end_date"] = pd.to_datetime(subscriptions["end_date"])

feature_usage["usage_date"] = pd.to_datetime(feature_usage["usage_date"])

support_tickets["submitted_at"] = pd.to_datetime(support_tickets["submitted_at"])
support_tickets["closed_at"] = pd.to_datetime(support_tickets["closed_at"])

churn_events["churn_date"] = pd.to_datetime(churn_events["churn_date"])
```

```python
date_columns = {
    "Accounts": accounts["signup_date"],
    "Subscriptions - start": subscriptions["start_date"],
    "Subscriptions - end": subscriptions["end_date"],
    "Feature Usage": feature_usage["usage_date"],
    "Support Tickets - submitted": support_tickets["submitted_at"],
    "Support Tickets - closed": support_tickets["closed_at"],
    "Churn Events": churn_events["churn_date"]
}

for name, column in date_columns.items():
    print(name, "→", column.dtype)
```
